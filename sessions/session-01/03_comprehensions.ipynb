# Session 01 — Segment C: Comprehensions

A comprehension builds a new list (or dict) by applying an expression to every item in an iterable — in one line. It replaces the three-line for-loop-plus-append pattern.

**Analogy**: A comprehension is a filtering and transforming pipeline on a conveyor belt — items come in, you apply a rule, only the ones that pass come out the other end as a new list.

## List comprehensions

```python
exercise_sets = [
    {"weight_kg": 80, "reps": 5},
    {"weight_kg": 0,  "reps": 10},  # bodyweight — no weight
    {"weight_kg": 82.5, "reps": 3},
]

# The for-loop way (3 lines):
weights = []
for s in exercise_sets:
    weights.append(s["weight_kg"])
print("loop:", weights)

# List comprehension — same result, 1 line:
weights = [s["weight_kg"] for s in exercise_sets]
print("comp:", weights)
```

```python
# With a filter — only sets where weight > 0
weights = [s["weight_kg"] for s in exercise_sets if s["weight_kg"] > 0]
print(weights)  # [80, 82.5] — the 0-weight set is excluded
```

Structure: `[expression for item in iterable if condition]`  
The `if condition` part is optional — it filters what comes through.

```python
# Transformation
names = ["riya", "arjun", "priya"]
print([name.upper() for name in names])           # ["RIYA", "ARJUN", "PRIYA"]
print([name.upper() for name in names if len(name) > 4])  # ["ARJUN", "PRIYA"]
```

## Dict comprehensions

```python
lifts = ["bench", "squat", "deadlift"]
weights = [80, 120, 140]

# Build a dict from two lists
pr_map = {lift: weight for lift, weight in zip(lifts, weights)}
print(pr_map)

# Transform values
pr_map_kg = {lift: f"{weight}kg" for lift, weight in pr_map.items()}
print(pr_map_kg)
```

## Generator expressions

```python
# Generator expression — same syntax as list comp but with () or used directly in a function
# Produces values one at a time instead of building the whole list in memory

# Inside join() — no intermediate list created
sets_str = " | ".join(
    f"{s['weight_kg']}kg×{s['reps']}"
    for s in exercise_sets
    if s["weight_kg"] > 0
)
print(sets_str)  # "80kg×5 | 82.5kg×3"
```

```python
# Inside sum(), any(), all()
total_volume = sum(s["weight_kg"] * s["reps"] for s in exercise_sets)
has_heavy    = any(s["weight_kg"] > 100 for s in exercise_sets)
all_logged   = all(s["weight_kg"] >= 0 for s in exercise_sets)

print(f"total volume: {total_volume}kg")
print(f"has heavy set: {has_heavy}")
print(f"all logged: {all_logged}")
```

## All patterns together — 10 lines

Every pattern from this session combined: `.get()`, f-strings, generator expression inside `join()`, list filtering.

```python
workouts = [
    {
        "start_time": "2026-06-01T09:00:00",
        "title": "Push Day",
        "exercises": [
            {"title": "Bench Press", "sets": [{"weight_kg": 80, "reps": 5}, {"weight_kg": 82.5, "reps": 3}]},
            {"title": "Overhead Press", "sets": [{"weight_kg": 0, "reps": 10}]},
        ]
    }
]

lines = ["# Recent Workouts\n"]
for w in workouts:
    date = w.get("start_time", "")[:10]         # .get() + slicing
    lines.append(f"\n[{date}] {w.get('title', 'Workout')}")
    for ex in w.get("exercises", []):
        sets_str = " | ".join(                  # join + generator expression
            f"{s['weight_kg']}kg×{s['reps']}"
            for s in ex.get("sets", [])
            if s.get("weight_kg")               # filter: skip 0-weight sets
        )
        if sets_str:
            lines.append(f"  {ex['title']}: {sets_str}")

print("\n".join(lines))
```

⚡ **Best practice**: Use generator expressions inside `join()`, `sum()`, `any()`, `all()` — you only iterate once and no intermediate list is built in memory.

**Misconception**: A list comprehension always builds the complete list in memory before returning. A generator expression produces values one at a time — use it when you only need to iterate once. For large datasets (80 workouts, 1000 sets), the difference matters.
