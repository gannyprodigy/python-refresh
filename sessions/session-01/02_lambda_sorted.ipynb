# Session 01 — Segment B: Lambda and `sorted()` with `key=`

A lambda is a tiny, inline function written in one expression — no `def`, no name, no `return` statement. You'll almost always see it as the `key=` argument inside `sorted()`, telling it *what to sort by*.

**Analogy**: A lambda is like a sticky note with one instruction. You wouldn't frame it — it's used once, right there, and thrown away.

```python
# sorted() with key= — the pattern in production
workouts = [
    {"name": "bench press", "max_weight": 80},
    {"name": "squat",       "max_weight": 120},
    {"name": "deadlift",    "max_weight": 140},
]

result = sorted(workouts, key=lambda w: w["max_weight"], reverse=True)
for r in result:
    print(r)
```

`key=lambda w: w["max_weight"]` tells `sorted()`: for each item `w`, extract `w["max_weight"]` and sort by that value. `reverse=True` gives descending order.

```python
# The lambda is equivalent to writing this named function:
def by_weight(w):
    return w["max_weight"]

result = sorted(workouts, key=by_weight, reverse=True)
for r in result:
    print(r)
```

Identical result. Lambda is just shorter when the function is trivial and used only once.

```python
# From real production code — sort personal records by weight, take top 10
prs = {"bench press": 80.0, "squat": 120.0, "deadlift": 140.0, "overhead press": 60.0}

for lift, weight in sorted(prs.items(), key=lambda x: -x[1])[:3]:
    print(f"{lift}: {weight}kg")
```

`prs.items()` produces pairs like `("bench press", 80.0)`. The lambda `lambda x: -x[1]` sorts by the second element (the weight) as a negative — which gives descending order without `reverse=True`. `[:3]` takes the top 3.

```python
# Sorting by multiple criteria: first by age, then alphabetically by name
patients = [
    {"name": "Riya",  "age": 34},
    {"name": "Arjun", "age": 34},
    {"name": "Priya", "age": 28},
]

result = sorted(patients, key=lambda p: (p["age"], p["name"]))
for r in result:
    print(r)
```

Returning a tuple from `key=` sorts by the first element first, then the second as a tiebreaker.

⚡ **Best practice**: Use lambda for simple one-expression keys. If the logic is more than one expression, write a proper `def` function — it's more readable and debuggable.

**Misconception**: Lambda can only contain one expression — no `if/else` blocks spanning multiple lines, no assignments. `lambda x: x["age"] if x["age"] > 0 else 0` is fine (ternary in one expression). A multi-line if-block is not.
