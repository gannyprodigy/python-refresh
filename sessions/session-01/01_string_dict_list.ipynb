# Session 01 — Segment A: String, List, Dict Methods

Python's built-in types come with methods attached. You call them with dot notation. A small set of them appears in every production AI file — this segment covers exactly those ones.

## String methods

```python
name = "  Riya Sharma  "
print(name.strip())           # removes leading/trailing whitespace
print(name.strip().split(" ")) # split on spaces → list of words
```

`.strip()` cleans whitespace — incoming API data almost always has it. `.split()` breaks a string into a list on a delimiter.

```python
conditions = ["diabetes", "asthma", "hypertension"]
print(", ".join(conditions))          # joins list items into one string
print(" | ".join(["80kg×5", "80kg×5"])) # any separator works
```

`.join()` is the inverse of `.split()`. The separator string calls `.join()` on a list. This appears constantly when building readable output from lists.

```python
# f-strings — the standard way to format strings in Python 3.12
patient = "Riya"
age = 34
weight = 85.5

print(f"Patient: {patient}, age {age}")
print(f"Weight: {weight:.1f}kg")   # .1f = 1 decimal place
print(f"Weight: {weight:.2f}kg")   # .2f = 2 decimal places
```

`f"{variable}"` drops the variable's value inline. `{value:.1f}` controls formatting — the colon introduces a format spec, `f` means float, `1` is decimal places.

## List methods

```python
meds = ["metformin", "lisinopril"]
meds.append("aspirin")                    # adds one item to the end
meds.extend(["vitamin D", "B12"])         # adds all items from another list
print(meds)
print(len(meds))                          # count of items
```

```python
scores = [4.2, 4.8, 3.9, 4.5]
print(sorted(scores))               # ascending — new list, original unchanged
print(sorted(scores, reverse=True)) # descending
print(scores)                       # original untouched
```

`sorted()` always returns a **new** list — the original is unchanged. `.sort()` (called on the list itself) modifies in place. In production code, `sorted()` is preferred because it doesn't mutate the input.

## Dict methods — the most important ones

```python
record = {"name": "Riya", "age": 34, "conditions": ["diabetes"]}

# Direct access — crashes if key missing
print(record["name"])      # "Riya" ✅
# record["weight"]         # KeyError: 'weight' ❌
```

```python
# .get() — safe access, never crashes
print(record.get("name"))           # "Riya"
print(record.get("weight"))         # None — key missing, no crash
print(record.get("weight", 0))      # 0 — key missing, use default
print(record.get("conditions", [])) # ["diabetes"] — or [] if missing
```

⚡ **Best practice**: Use `.get("key", default)` on any dict from an external source — API responses, webhook payloads, database rows. A missing key from an API would crash your whole request handler with `dict["key"]`. `.get()` never crashes.

```python
# Iterating over a dict
for key, value in record.items():
    print(f"{key}: {value}")

print(list(record.keys()))    # all keys as a list
print(list(record.values()))  # all values as a list
```

```python
# Building a dict from two lists with zip()
lifts = ["bench", "squat", "deadlift"]
weights = [80, 120, 140]

pr_map = dict(zip(lifts, weights))
print(pr_map)  # {"bench": 80, "squat": 120, "deadlift": 140}
```

## The one misconception

`dict["key"]` crashes with `KeyError` if the key doesn't exist. `dict.get("key", default)` never crashes — it returns the default. In production code, every dict from an external API uses `.get()`. A missing key from an API silently kills the request without it.
